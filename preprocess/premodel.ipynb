# 数据集统计分析

```python
import json
import ijson 

sample_1xx = []
sample_2xx = []
all_sample = []

# 打开samples_1xx.jsonl和samples_2xx.jsonl文件并逐行读取
# 两个文件的所有内容都放到all_sample列表中
with open('../DATAsets/samples_1xx.jsonl', 'r', encoding='utf-8') as f1:
    for line in f1:
        item = json.loads(line)
        sample_1xx.append(item)
with open('../DATAsets/samples_2xx.jsonl', 'r', encoding='utf-8') as f2:
    for line in f2:
        item = json.loads(line)
        sample_2xx.append(item)

all_sample = sample_1xx + sample_2xx
```

```python
# 输出keys
print(all_sample[0].keys())
print(f"Total samples loaded: {len(all_sample)}")
```

## 设备信息和样本类型

```python
# 统计brand、type和product的分布
from collections import Counter

brand_counts = Counter(item.get('brand', 'Unknown') for item in all_sample)
type_counts = Counter(item.get('type', 'Unknown') for item in all_sample)
product_counts = Counter(item.get('product', 'Unknown') for item in all_sample)
print("Brand distribution:", brand_counts)
print("Type distribution:", type_counts)
print("Product distribution:", product_counts)
```

```python
# 有多少不重复的brand、type和product
unique_brands = set(item.get('brand') for item in all_sample if item.get('brand'))
unique_types = set(item.get('type') for item in all_sample if item.get('type'))
unique_products = set(item.get('product') for item in all_sample if item.get('product'))

print(f"共有品牌: {len(unique_brands)}种")
print(f"共有类型: {len(unique_types)}种")
print(f"共有产品: {len(unique_products)}种")
```

修改字段名

```python
import json
import os

def rename_field_in_jsonl(file_path, old_field, new_field):
    """
    将JSONL文件中的指定字段重命名为新字段。
    """
    temp_file = file_path + '.tmp'
    with open(file_path, 'r', encoding='utf-8') as infile, open(temp_file, 'w', encoding='utf-8') as outfile:
        for line in infile:
            data = json.loads(line.strip())
            if old_field in data:
                data[new_field] = data.pop(old_field)
            outfile.write(json.dumps(data, ensure_ascii=False) + '\n')
    os.replace(temp_file, file_path)
    print(f"Processed {file_path}")

# 示例：处理多个数据集文件
files_to_process = [
    '../DATASETs/final_test_set.jsonl'
    # '../DATAsets/sampled_0.1p.jsonl',
    # '../DATAsets/sampled_0.5p.jsonl',
    # '../DATAsets/sampled_1p.jsonl',
    # '../DATAsets/sampled_5p.jsonl',
    # '../DATAsets/sampled_8p.jsonl'
    # 添加其他需要处理的JSONL文件路径
]

for file_path in files_to_process:
    if os.path.exists(file_path):
        rename_field_in_jsonl(file_path, 'text', 'banner')
    else:
        print(f"File not found: {file_path}")
```

## 筛选测试集

```python
# 正样本 8967 条（14014×64.7%）、负样本 5047 条（14014×35.3%）
import jsonlines
import json

# 读取samples_1xx.jsonl和samples_2xx.jsonl
samples_1xx = []
samples_2xx = []
with open('../DATAsets/samples_1xx.jsonl', 'r', encoding='utf-8') as f1:
    for line in f1:
        item = json.loads(line)
        samples_1xx.append(item)

with open('../DATAsets/samples_2xx.jsonl', 'r', encoding='utf-8') as f2:
    for line in f2:
        item = json.loads(line)
        samples_2xx.append(item)
```

```python
# 过滤samples_1xx和samples_2xx中在sampled_8p.jsonl中的样本
filtered_samples_1xx = []
filtered_samples_2xx = []
with open('../DATAsets/sampled_8p_cleaned.jsonl', 'r', encoding='utf-8') as f:
    sampled_8p = {json.loads(line)['id'] for line in f}

for item in samples_1xx:
    if item['id'] not in sampled_8p:
        filtered_samples_1xx.append(item)

for item in samples_2xx:
    if item['id'] not in sampled_8p:
        filtered_samples_2xx.append(item)
print(f"Filtered samples_1xx: {len(filtered_samples_1xx)}")
print(f"Filtered samples_2xx: {len(filtered_samples_2xx)}")
```

```python
# 在filtered_samples_1xx中抽样8967条，按照设备类型分层抽样
import random
random.seed(42)  # 设置随机种子以确保可重复性

# 统计各设备类型的样本数量
device_counts = {}
for item in filtered_samples_1xx:
    device = item['type']
    if device not in device_counts:
        device_counts[device] = 0
    device_counts[device] += 1

# 按照设备类型分层抽样
stratified_samples = []
for device, count in device_counts.items():
    # 计算该设备类型需要抽样的数量
    num_samples = int(9005 * count / len(filtered_samples_1xx))
    # 从该设备类型中随机抽样
    device_samples = [item for item in filtered_samples_1xx if item['type'] == device]
    stratified_samples.extend(random.sample(device_samples, min(num_samples, len(device_samples))))


print(f"Stratified sampled positive samples: {len(stratified_samples)}")
```

```python
# 在filtered_samples_2xx中抽样4885条，按照设备类型分层抽样
import random
random.seed(42)  # 设置随机种子以确保可重复性

# 统计各设备类型的样本数量
device_counts = {}
for item in filtered_samples_2xx:
    device = item['type']
    if device not in device_counts:
        device_counts[device] = 0
    device_counts[device] += 1

# 按照设备类型分层抽样
for device, count in device_counts.items():
    # 计算该设备类型需要抽样的数量
    num_samples = int(4885 * count / len(filtered_samples_2xx))
    # 从该设备类型中随机抽样
    device_samples = [item for item in filtered_samples_2xx if item['type'] == device]
    stratified_samples.extend(random.sample(device_samples, min(num_samples, len(device_samples))))


print(f"Stratified sampled negative samples: {len(stratified_samples)}")
```

```python
# 保存为final_test_set.jsonl
with jsonlines.open('../DATAsets/final_test_set.jsonl', 'w') as writer:
    for item in stratified_samples:
        writer.write(item)
```
